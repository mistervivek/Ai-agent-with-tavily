```python
import os
import getpass
from dotenv import load_dotenv
from tavily import TavilyClient

load_dotenv()

if not os.environ.get("TAVILY_API_KEY"):
    os.environ["TAVILY_API_KEY"] = getpass.getpass("TAVILY_API_KEY:\n")

tavily_client = TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))
```

```python
search_results = tavily_client.search(
    query="What happened in NYC today?", max_results=5
)
```

```python
### Print the results
for result in search_results["results"]:
    print(result["title"])
    print(result["url"])
    print(result["content"])
    print(result["score"])
    print("\n")
```

```python
search_results = tavily_client.search(
    query="Anthropic model release?",
    max_results=5,
    time_range="month",
    include_domains=["techcrunch.com"],
    topic="news",
)
```

```python
for result in search_results["results"]:
    print(result["title"])
    print(result["url"])
    print(result["content"])
    print("\n")
```

```python
extract_results = tavily_client.extract(
    urls=[result["url"] for result in search_results["results"]],
)
```

```python

for result in extract_results["results"]:
    print(result["url"])
    print(result["raw_content"])
    print("\n")
```

```python
search_results = tavily_client.search(
    query="Anthropic model release?",
    max_results=1,
    include_raw_content=True,
)
```

```python

for result in search_results["results"]:
    print(result["url"])
    print(result["content"])
    print(result["score"])
    print(result["raw_content"])
    print("\n")
```

```python
crawl_results = tavily_client.crawl(url="tavily.com")
```

```python
for result in crawl_results["results"]:
    print(result["url"])
```

```python

for result in crawl_results["results"]:
    print(result["url"])
    print(result["raw_content"])
    print("\n")
```
